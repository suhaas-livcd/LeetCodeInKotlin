- Time Complexity: O(n). We are iterating through each string once.
- Space Complexity: O(k). We don't count the output as part of the space complexity, but for each word, we are using some space for the length and delimiter.

```kotlin
class Codec {
    // Encodes a list of strings to a single string.
    fun encode(strs: List<String>): String {
        val sb = StringBuilder()
        strs.forEach{
            sb.append(it.length).append("#").append(it)
        }
        return sb.toString()
    }

    // Decodes a single string to a list of strings.
    fun decode(s: String): List<String> {
        var i = 0
        val result: MutableList<String> = mutableListOf()
        while(i < s.length){
            var j = i
            // get the size of the word
            while(s[j] != '#') j++
            val wordLen = s.substring(i, j).toInt()
            val start = j + 1
            val end = start + wordLen
            result.add(s.substring(start, end))
            i = end
        }
        return result
    }
}
```
